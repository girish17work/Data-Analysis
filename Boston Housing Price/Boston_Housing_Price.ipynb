```python
X = df.drop(columns=['Category', 'PRICE'])
y = df['PRICE']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

min_max_cols = ['CRIM', 'ZN', 'POLINDEX', 'DIS', 'HIGHWAYCOUNT', 'TAX', 'PTRATIO', 'IMM', 'BPL']
standard_scale_cols = ['INDUS']
rm_cols = ['RM']
age_cols = ['AGE']
riverside_cols = ['RIVERSIDE']

rm_pipeline = Pipeline([('imputer', SimpleImputer(strategy='median')), ('scaler', MinMaxScaler())])
age_pipeline = Pipeline([('imputer', SimpleImputer(strategy='mean')), ('scaler', MinMaxScaler())])
riverside_pipeline = Pipeline([('imputer', SimpleImputer(strategy='most_frequent')), ('encoder', OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore'))])

preprocessor = ColumnTransformer(
    transformers=[
        ('min_max', MinMaxScaler(), min_max_cols),
        ('standard_scaler', StandardScaler(), standard_scale_cols),
        ('rm', rm_pipeline, rm_cols),
        ('age', age_pipeline, age_cols),
        ('riverside', riverside_pipeline, riverside_cols)
    ],
    remainder='passthrough'
).set_output(transform='pandas')

X_train_preprocessed = preprocessor.fit_transform(X_train)
X_test_preprocessed = preprocessor.transform(X_test)
```

```python
print(f"The number of features after performing transformation is {X_train_preprocessed.shape[1]}")
```

```python
print(f"The mean of the tranformed test data (features only) is {round(X_test_preprocessed.to_numpy().mean(), 3)}")
```
