```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

```python
DATASET_URL = 'https://drive.google.com/uc?export=download&id=14ZMZFo6XHGeaZY1mTOL9GBwjLvSW4suR'
df = pd.read_csv(DATASET_URL)
df.head()
```

```python
print(f'Unique values present in the Residential feature: {df['Residential'].unique()}')
```

```python
print(f'Columns with categorical data: {df.select_dtypes(include=['object']).columns.tolist()}')
```

```python
print(f'Features with missing values: {df.columns[df.isna().any()].to_list()}')
```

```python
print(f'Features with unknown values: {df.columns[(df == '?').any()].to_list()}')
```

```python
print(f'In the year 2021, the number of houses in Stamford locality having 2 or more num_bathrooms and are facing towards either the South or West is {((df['Year']==2021) & (df['Locality']=='Stamford') & (df['num_bathrooms']>=2) & (df['Face'].isin(['South', 'West']))).sum()}')
```

```python
train_df = df[df['Year'].between(2009, 2019)]
test_df = df[df['Year'].isin([2020, 2021, 2022])]
X_train = train_df.drop(columns=['Sale Price'])
y_train = train_df['Sale Price']
X_test = test_df.drop(columns=['Sale Price'])
y_test = test_df['Sale Price']

print(f'The number of rows in the feature matrix of the train dataset: {X_train.shape[0]}')
```

```python
locality_mode = X_train['Locality'].mode()[0]
estimated_value_mean = X_train['Estimated Value'].mean()
carpet_area_median = X_train['carpet_area'].median()
property_mode = X_train['Property'].mode()[0]

for df in [X_train, X_test]:
  df['Locality'] = df['Locality'].fillna(locality_mode)
  df['Estimated Value'] = df['Estimated Value'].fillna(estimated_value_mean)
  df['carpet_area'] = df['carpet_area'].fillna(carpet_area_median)
  df['Property'] = df['Property'].replace('?', property_mode)
```

```python
print(f'Median of the carpet_area column of the train dataset: {X_train['carpet_area'].median()}')
```

```python
print(f'Value used to replace missing values in the Locality column: {locality_mode}')
```

```python
X_train = X_train.drop(columns=['Date'])
X_test = X_test.drop(columns=['Date'])
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

numerical_features = X_train.select_dtypes(include=['number']).columns.to_list()
categorical_features = X_train.select_dtypes(include=['object']).columns.to_list()

preprocessor = ColumnTransformer(transformers=[
    ('ohe', OneHotEncoder(sparse_output=False, handle_unknown='ignore'), categorical_features),
    ('scaler', StandardScaler(), numerical_features)
]).set_output(transform='pandas')

X_train = preprocessor.fit_transform(X_train)
X_test = preprocessor.transform(X_test)
```

```python
print(f'Total features after one hot encoding and scaling is {X_train.shape[1]}')
```

```python
print(f'Maximum value present in the Test Feature Matrix: {round(X_test.to_numpy().max(), 2)}')
```

```python
DATASET_URL = 'https://drive.google.com/uc?export=download&id=19C-f3TvlbYeOBubiNsk-yFuUux_ENQxh'
df = pd.read_csv(DATASET_URL)
df.head()
```

```python
train_df = df[:8000]
test_df = df[8000:]
X_train = train_df.drop(columns=['Sale Price'])
X_test = test_df.drop(columns=['Sale Price'])
y_train = train_df['Sale Price']
y_test = test_df['Sale Price']
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print(f'Mean Absolute Error: {round(mean_absolute_error(y_test, y_pred), 2)}')
```

```python
from numpy import savez_compressed
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LinearRegression

estimator = LinearRegression()
scores = cross_val_score(estimator, X_train, y_train, cv=5)
print(f'Minimum Score: {round(scores.min(), 4)}')
```

```python
from sklearn.linear_model import Ridge, Lasso
from sklearn.metrics import mean_absolute_error

ridge = Ridge(random_state=27)
lasso = Lasso(random_state=27)

ridge.fit(X_train, y_train)
lasso.fit(X_train, y_train)

y_pred_ridge = ridge.predict(X_test)
y_pred_lasso = lasso.predict(X_test)

ridge_score = mean_absolute_error(y_test, y_pred_ridge)
lasso_score = mean_absolute_error(y_test, y_pred_lasso)

print(f'Ridge Score: {ridge_score}')
print(f'Lasso Score: {lasso_score}')
```

```python
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import r2_score

sgd_regressor = SGDRegressor(random_state=27, warm_start=True, max_iter=50)
sgd_regressor.fit(X_train, y_train)
y_pred = sgd_regressor.predict(X_test)

print(f'R2 score: {round(r2_score(y_test, y_pred), 4)}')
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Lasso
from sklearn.model_selection import GridSearchCV

pipeline = Pipeline([('poly', PolynomialFeatures()), ('lasso', Lasso())])
param_grid = {
    'poly__degree': [1,2],
    'lasso__alpha': [1, 10, 100, 1000]
}
grid_search_cv = GridSearchCV(pipeline, param_grid=param_grid, scoring='neg_mean_squared_error', cv=5, n_jobs=-1)
grid_search_cv.fit(X_train, y_train)
```

```python
print(f'Best alpha value: {grid_search_cv.best_params_['lasso__alpha']}')
```

```python
print(f'Best polynomial degree: {grid_search_cv.best_params_['poly__degree']}')
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=10, svd_solver='full', whiten=True, random_state=27)
X_train_pca = pca.fit_transform(X_train)
X_test_pca = pca.transform(X_test)
```

```python
print(f'Sum of explained variance ratio: {round(pca.explained_variance_ratio_.sum(), 2)}')
```

```python
from sklearn.linear_model import RidgeCV
from sklearn.metrics import r2_score

ridge_cv = RidgeCV(alphas=[0.01, 0.1, 1, 10, 100, 1000], cv=5)
ridge_cv.fit(X_train_pca, y_train)

y_pred = ridge_cv.predict(X_test_pca)
print(f'R2 Score: {round(r2_score(y_test, y_pred), 4)}')
```

```python
from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.linear_model import Lasso
from sklearn.pipeline import Pipeline

pipeline = Pipeline([('kbest', SelectKBest(f_regression, k=8)), ('lasso', Lasso(random_state=27))])
pipeline.fit(X_train, y_train)
```

```python
kbest = pipeline.named_steps['kbest']
print(f'Number of features on the train dataset that has p value less than 0.05: {(kbest.pvalues_<0.05).sum()}')
```

```python
print(f'Minimum F Statistics score for any feature on the train dataset: {(round(kbest.scores_.min(), 2))}')
```

```python
from sklearn.metrics import mean_absolute_error

y_pred = pipeline.predict(X_test)
print(f'Mean Absolute Error for the test dataset: {round(mean_absolute_error(y_test, y_pred), 2)}')
```
