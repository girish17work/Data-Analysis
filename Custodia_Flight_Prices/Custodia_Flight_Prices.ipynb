```python
import pandas as pd
import numpy as np
```

```python
DATASET_URL = 'https://custodia-beta.vercel.app/datasets/mlp/flight_prices_raw.csv'
df = pd.read_csv(DATASET_URL)
df.head()
```

```python
print(f'Mean flight ticket price: {round(df['Price'].mean(), 2)}')
```

```python
print(f'Maximum flight ticket price: {df['Price'].max()}')
```

```python
print(f'Unique Airline: {df['Airline'].nunique()}')
```
