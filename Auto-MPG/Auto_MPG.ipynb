# Auto MPG Dataset

```python
import pandas as pd
```

```python
column_names = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model_year', 'origin', 'car_name']
auto_mpg_dataset = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/auto-mpg/auto-mpg.data', sep=r'\s+', header=None, names=column_names)
```

```python
print(f"The dataset has {auto_mpg_dataset.shape[0]} samples and {auto_mpg_dataset.shape[1]} features")
```
